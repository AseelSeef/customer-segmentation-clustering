drop duplicates

```python
customer_data.drop_duplicates(inplace=True)

print(f"view new data shape : \n{customer_data.shape}")
print(f"view new data frame : \n{customer_data.head()}")
```

```python
columns_to_keep = [
    "Delivery_Date",
    "Total Revenue",
    "Unit Cost",
    "Customer_BirthDate",
    "Order ID",
    "Customer ID"
]

cleaned_df = customer_data[columns_to_keep]
```

```python
print(f"view cleaned data shape : \n{cleaned_df.shape}")
print(f"view cleaned data frame : \n{cleaned_df.head()}")
cleaned_df.head().T
```

```python
# Frequency - number of orders per customer
cleaned_df['Frequency'] = cleaned_df.groupby('Customer ID')['Order ID'].transform('count')

# Recency - days since last purchase
cleaned_df['Delivery_Date'] = pd.to_datetime(cleaned_df['Delivery_Date'])
cleaned_df['Recency'] = (pd.Timestamp.today() - cleaned_df['Delivery_Date']).dt.days

# Age calculation
cleaned_df['DOB'] = pd.to_datetime(cleaned_df['Customer_BirthDate'], format='%d%b%Y', errors='coerce')
cleaned_df['Age'] = ((pd.Timestamp.today() - cleaned_df['DOB']).dt.days / 365).round(0)

# Clean Total Revenue and convert to float
cleaned_df['CLV'] = (
    cleaned_df['Total Revenue']
    .astype(str)
    .str.replace(r'[ $,]', '', regex=True)
    .astype(float)
)

# Clean Unit Cost and convert to float
cleaned_df['UnitCost'] = (
    cleaned_df['Unit Cost']
    .astype(str)
    .str.replace(r'[ $,]', '', regex=True)
    .astype(float)
)
```

outliers analysis

```python
def calc_iqr():
    """
    Function to calculate IQR for each feature
    and create binary outlier columns.
    """
    for col in df_columns:
        q1 = customer_df[col].quantile(0.25)
        q3 = customer_df[col].quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        outliers = customer_df[(customer_df[col] < lower) | (customer_df[col] > upper)]
        percentage = len(outliers) / len(customer_df) * 100
        customer_df[f"{col}_outlier"] = ((customer_df[col] < lower) | (customer_df[col] > upper)).astype(int)

        print(col, "outliers:", len(outliers))
        print("Percentage:", percentage)
        print("\n")

    #identify the number of features that in outlier condition --- threshold
    outlier_cols = [f"{col}_outlier" for col in df_columns]
    customer_df["num_outlier_features"] = customer_df[outlier_cols].sum(axis=1)
    total = len(customer_df)
    for t in range(1, 7):
        percent = (customer_df["num_outlier_features"] >= t).sum() / total * 100
        print(f"threshold {t} -> {percent:.2f}%")
```

```python
calc_iqr()
```

Using the IQR method, the dataset shows moderate variability with a small proportion of outliers across most features. These outliers likely represent natural variation in customer behavior rather than data errors, and should be evaluated in the context of business meaning before removal.

# Clustering Analysis

## Elbow method

```python
# Elbow method: try different k.
wcss=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i, init='k-means++',random_state=0, n_init = 10)
    kmeans.fit(df_preprocessed)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,11),wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

The Elbow Method was used to determine the optimal number of clusters by analyzing the WCSS curve. A sharp decrease in WCSS is observed up to k = 4, after which the rate of decrease slows significantly. This indicates that adding more clusters beyond this point provides diminishing returns. Therefore, the optimal number of clusters is chosen to be between 4 and 5.

## Silhouette score method

```python
import matplotlib.cm as cm

range_n_clusters = [4, 5]
X = df_preprocessed.values

for n_clusters in range_n_clusters:
    # Create a subplot with 1 row and 2 columns.
    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(18, 7)

    # The 1st subplot is the silhouette plot.
    # The silhouette coefficient can range from -1, 1 but in this example all
    # lie within [-0.1, 1].
    ax1.set_xlim([-0.1, 1])
    # The (n_clusters+1)*10 is for inserting blank space between silhouette
    # plots of individual clusters, to demarcate them clearly.
    ax1.set_ylim([0, len(X) + (n_clusters + 1) * 10])

    # Initialize the clusterer with n_clusters value and a random generator
    # seed of 10 for reproducibility.
    clusterer = KMeans(n_clusters=n_clusters, random_state=10, n_init = 10)
    cluster_labels = clusterer.fit_predict(X)

    # The silhouette_score gives the average value for all the samples.
    # This gives a perspective into the density and separation of the formed
    # clusters
    silhouette_avg = silhouette_score(X, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette scores for each sample
    sample_silhouette_values = silhouette_samples(X, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        # Aggregate the silhouette scores for samples belonging to
        # cluster i, and sort them.
        ith_cluster_silhouette_values = \
            sample_silhouette_values[cluster_labels == i]

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        # Label the silhouette plots with their cluster numbers at the middle.
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        # Compute the new y_lower for next plot.
        y_lower = y_upper + 10  # 10 for the 0 samples.

    ax1.set_title("The silhouette plot for the various clusters.")
    ax1.set_xlabel("The silhouette coefficient values")
    ax1.set_ylabel("Cluster label")

    # The vertical line for average silhouette score of all the values.
    ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

    ax1.set_yticks([])  # Clear the yaxis labels / ticks.
    ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

    # 2nd Plot showing the actual clusters formed.
    colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)
    ax2.scatter(X[:, 0], X[:, 1], marker='.', s=30, lw=0, alpha=0.7,
                c=colors, edgecolor='k')

    # Labeling the clusters.
    centers = clusterer.cluster_centers_
    # Draw white circles at cluster centers.
    ax2.scatter(centers[:, 0], centers[:, 1], marker='o',
                c="white", alpha=1, s=200, edgecolor='k')

    for i, c in enumerate(centers):
        ax2.scatter(c[0], c[1], marker='$%d$' % i, alpha=1,
                    s=50, edgecolor='k')

    ax2.set_title("The visualization of the clustered data.")
    ax2.set_xlabel("Feature space for the 1st feature")
    ax2.set_ylabel("Feature space for the 2nd feature")

    plt.suptitle(("Silhouette analysis for KMeans clustering on sample data "
                  "with n_clusters = %d" % n_clusters),
                 fontsize=14, fontweight='bold')

plt.show()
```

The average silhouette score for k = 5 (0.267) was slightly higher than for k = 4 (0.253), indicating better-defined clusters.
The silhouette plot for k = 5 shows that most data points have positive silhouette coefficients, indicating that they are well matched to their assigned clusters.
So optimal number of clusters is k=5

## Cluster Profiling & Interpretation

```python
# Function to perform k-means clustering and plot the clusters in 2D.
def plot_k_means(data, features, scaled=False):
    # Extract the selected features
    X = data[features].values

    # Optionally scale the features.
    if scaled:
        scaler = StandardScaler()
        X = scaler.fit_transform(X)

    # Perform k-means clustering.
    kmeans = KMeans(n_clusters=4, random_state=42,  n_init=10)
    kmeans.fit(X)
    labels = kmeans.labels_

    # 2D scatter plot.
    plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')

    # Plotting centroids.
    centroids = kmeans.cluster_centers_
    plt.scatter(centroids[:, 0],
                centroids[:, 1],
                s=100, c='red',
                marker='X')

    # Labeling axes.
    plt.xlabel(features[0])
    plt.ylabel(features[1])

    # Title.
    if scaled:
        plt.title('K-means clustering with scaled features (2D)')
    else:
        plt.title('K-means clustering with unscaled features (2D)')

    # Show plot.
    plt.show()

# Choose two features for 2D plotting.
selected_features = ['Recency', 'Frequency']

# Plot unscaled data.
plot_k_means(df_preprocessed, selected_features, scaled=False)
```

## PCA

```python
def plot_pca_clusters(data, cluster_labels, title):
    """
    Applies PCA to reduce data to 2D and plots clusters.
    """
    # Apply PCA
    pca = PCA(n_components=2)
    x_pca = pca.fit_transform(data)

    # Create a new DataFrame
    df_pca = pd.DataFrame({
        "PCA1": x_pca[:, 0],
        "PCA2": x_pca[:, 1],
        "Cluster": cluster_labels
    })

    # Plot
    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(
        df_pca["PCA1"],
        df_pca["PCA2"],
        c=df_pca["Cluster"],
        cmap="coolwarm",
        alpha=0.7
    )
    plt.title(title)
    plt.xlabel("PCA1")
    plt.ylabel("PCA2")
    plt.colorbar(scatter, label="Cluster")
    plt.show()
```

```python
from sklearn.manifold import TSNE

def plot_tsne_clusters(data, cluster_labels, title):
    """
    Applies t-SNE to reduce data to 2D and plots clusters.
    """

    tsne = TSNE(n_components=2, random_state=42, perplexity=30)
    x_tsne = tsne.fit_transform(data)

    df_tsne = pd.DataFrame({
        "TSNE1": x_tsne[:, 0],
        "TSNE2": x_tsne[:, 1],
        "Cluster": cluster_labels
    })

    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(
        df_tsne["TSNE1"],
        df_tsne["TSNE2"],
        c=df_tsne["Cluster"],
        cmap="coolwarm",
        alpha=0.7
    )

    plt.title(title)
    plt.xlabel("t-SNE 1")
    plt.ylabel("t-SNE 2")
    plt.colorbar(scatter, label="Cluster")
    plt.show()
```

```python
df_sample = df_preprocessed.sample(2000, random_state=42)
labels_sample = customer_df.loc[df_sample.index, "Cluster"]

plot_tsne_clusters(
    data=df_sample,
    cluster_labels=labels_sample,
    title="t-SNE Visualization of Customer Clusters"
)
```
